# 3. Symlink data files

```python
import os

DRIVE_DATA_PATH = "/content/drive/MyDrive/gasnet-data"

if not os.path.exists(DRIVE_DATA_PATH):
    os.makedirs(DRIVE_DATA_PATH)
    print(f"Created persistent folder at {DRIVE_DATA_PATH}")

datasets = ['data', 'checkpoints', 'results']

local_dataset_dir = "/content/gasnet"

for ds in datasets:
    drive_ds_path = os.path.join(DRIVE_DATA_PATH, ds)
    if not os.path.exists(drive_ds_path):
        os.makedirs(drive_ds_path)

    local_ds_path = os.path.join(local_dataset_dir, ds)

    # If the local folder already exists as a real directory, remove it first
    if os.path.exists(local_ds_path) and not os.path.islink(local_ds_path):
        !rm -rf {local_ds_path}

    # Link local path to Drive path
    if not os.path.islink(local_ds_path):
        os.symlink(drive_ds_path, local_ds_path)
        print(f"Linked local {ds} to Drive.")

print("Ready! Downloads will now save directly to your Google Drive.")
```
